```python
MODEL_ENDPOINT = 'http://iris-classifier-v1.models.svc.cluster.local'
MODEL_NAME = 'iris-classifier-v1'
```

```python
import json
from kserve import RESTConfig, InferenceRESTClient

config = RESTConfig(protocol="v2", retries=5, timeout=30)
client = InferenceRESTClient(config)

data = {
    'inputs': [
        {'name': 'sepal length (cm)',
           'shape': [3],
           'datatype': 'FP64',
           'data': [6, 4, 8]},
        {'name': 'petal length (cm)',
           'shape': [3],
           'datatype': 'FP64',
           'data': [3, 2, 7]}
    ]
}

result = await client.infer(MODEL_ENDPOINT, data, model_name=MODEL_NAME)

print(result.outputs[0].data)
```
